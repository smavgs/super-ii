```python
import os
import socket
from pathlib import Path

assert not any("SECRET" in key or key.endswith("_TOKEN") for key in os.environ)
try:
    Path("/workspace/forbidden-write").write_text("blocked")
except OSError:
    pass
else:
    raise AssertionError("repository mount is writable")
probe = socket.socket()
probe.settimeout(0.5)
try:
    probe.connect(("1.1.1.1", 53))
except OSError:
    pass
else:
    raise AssertionError("network egress is available")
finally:
    probe.close()
Path("/tmp/superii-probe").write_text("ok")  # noqa: S108
print("isolation verified")
```
