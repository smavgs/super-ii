# Start a Python project with Super ii

Use the released `superii-sdk` to inspect this machine and prepare a real repository workflow. The catalogue can truthfully be empty, so this notebook never substitutes a fake model. Install the exact tested release first: `python -m pip install superii-sdk==0.2.2`.

```python
import importlib.metadata
import os
import superii

version = importlib.metadata.version("superii-sdk")
if version != "0.2.2":
    raise RuntimeError(f"Expected superii-sdk 0.2.2, found {version}")
print(f"superii-sdk {version}")
```

## Inspect before acquiring

Hardware inspection stays on this computer. It is a planning input, not a benchmark or a guarantee that an arbitrary model will fit.

```python
machine = superii.hardware()
print(machine)
```

## Inspect a real reviewed repository

Set `SUPERII_REPOSITORY` to an actual `owner/slug` from the live catalogue. Inspection resolves metadata and immutable evidence only; it does not execute repository code.

```python
repository = os.getenv("SUPERII_REPOSITORY", "").strip()
manifest = None
plan = None
if not repository:
    print("Set SUPERII_REPOSITORY after choosing a real reviewed model.")
else:
    manifest = superii.inspect(repository)
    plan = superii.plan(manifest, machine=machine)
    print(manifest)
    print(plan)
```

## Make acquisition explicit

Downloading is separate from inspection. Keep the guard off until you have reviewed the exact revision, selected files, size, license, and plan. Every acquired file is verified by SHA-256 before use.

```python
ALLOW_DOWNLOAD = False
if ALLOW_DOWNLOAD and repository and plan is not None:
    snapshot = superii.pull(repository, revision=plan.revision, files=tuple(plan.files))
    if not superii.verify(snapshot):
        raise RuntimeError("Snapshot verification failed closed")
    print(snapshot.path)
else:
    print("No files downloaded. Review the plan, then set ALLOW_DOWNLOAD = True.")
```
