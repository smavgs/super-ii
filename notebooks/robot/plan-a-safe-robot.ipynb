# Plan a Robot without claiming physical safety

Ask the public deterministic planner for a source-backed Rover One starting plan. The result keeps verified, declared, derived, and unknown evidence separate and never grants physical safety approval.

```python
import json
from urllib.request import Request, urlopen

ORIGIN = "https://superii.site"
planner_input = {
    "goal": "learn",
    "experience": "new",
    "compute": "recommend",
    "environment": "indoor",
    "budget": "not-sure",
    "owned_components": [],
}
request = Request(
    f"{ORIGIN}/api/robot/plan",
    data=json.dumps(planner_input).encode("utf-8"),
    headers={"Content-Type": "application/json", "Accept": "application/json"},
    method="POST",
)
with urlopen(request, timeout=20) as response:
    plan = json.load(response)["plan"]
print(plan["title"], plan["catalog_revision"])
```

## Review evidence and unknowns

A component listing is not a guarantee that an assembled robot will be mechanically, electrically, thermally, or operationally safe. Empty compatibility evidence means unknown—not compatible.

```python
assert plan["safety_approval"] is False
states = {}
for component in plan["components"]:
    states[component["evidenceStatus"]] = states.get(component["evidenceStatus"], 0) + 1
print("Evidence states:", states)
print("Open requirements:")
for requirement in plan["open_requirements"]:
    print("-", requirement)
print("Robot Check:")
for check in plan["robot_check"]:
    print(f"- {check['dimension']}: {check['status']} — {check['explanation']}")
```

## Keep the physical boundary explicit

Before buying or energizing hardware, verify manufacturer documentation, voltage and current, connectors, cooling, mounting, payload and center of gravity, battery protection, emergency stop, environment, supervision, and local rules with qualified people.

```python
physical_review = {
    "manufacturer_sources_opened": False,
    "electrical_review_completed": False,
    "mechanical_review_completed": False,
    "thermal_review_completed": False,
    "battery_and_emergency_stop_reviewed": False,
    "qualified_human_approval": False,
}
print(json.dumps(physical_review, indent=2))
print("No purchase, actuator command, or hardware control was performed.")
```
